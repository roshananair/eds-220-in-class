```python
# Streamlined Workflow

# Import packages
import pandas as pd
import numpy as np

# Read in data
hare = pd.read_csv('55_Hare_Data_2012.txt')

# Sex dict
sex_dict = {'M' : 'male', 'm' : 'male', 'm ' : 'male', 'M?' : 'male', 'm?' : 'male', 'F' : 'female', 'f' : 'female', 'f ' : 'female', 'F?' : 'female','f?' : 'female', 'pf' : 'female', '?' : np.nan, 'nan' : np.nan}

# Replace values
hare['simple_sex'] = hare['sex'].map(sex_dict)

# Counts of hares by sex
print('Hare count by sex:')
print(hare['simple_sex'].value_counts(dropna = False))

# Mean hare weight by sex
print('Mean hare weight by sex:')
print(hare[['simple_sex', 'weight']].groupby('simple_sex').mean())
```
